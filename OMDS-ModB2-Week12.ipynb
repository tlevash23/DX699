# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data? The first chart borderline hurts to read in the traditional context and I would lean toward the second chart in terms of preference for visualizing this data. The Y-Axis on the first chart does not line up between the two graphs, this makes it hard to understand the trend on a simple level. There are only seven days in a week as well and the first chart is broken down in increments of 2 rather than showing each day individually. The second chart makes it very easy to see that in the beginning of one week they had more customers but as the week went on the number of visitors declined. On the other hand, the other week had a low number of visitors to start but then showed an increase in overall visitrs. The second chart is definitely easier for me to read because it is on a shared X-axis and the Y-axis is in the same increments across. 

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
#Chart from page 188
years = ["2008", "2009", "2010", "2011", "2012", "2013", "2014"]
products = ["Product A", "Product B", "Product C", "Product D", "Product E"]

values = np.array([
    [390, 410, 420, 390, 300, 270, 250],  # Product A
    [360, 400, 410, 370, 290, 260, 240],  # Product B
    [np.nan, np.nan, 100, 180, 200, 240, 180],  # Product C
    [np.nan, np.nan, np.nan, 150, 260, 220, 210],  # Product D
    [np.nan, np.nan, np.nan, np.nan, np.nan, 100, 205],  # Product E
])


x = np.arange(len(products))
width = 0.12  

fig, ax = plt.subplots(figsize=(10, 5))
for i, year in enumerate(years):
    offset = (i - (len(years)-1)/2) * width
    vals = values[:, i]
    
    mask = ~np.isnan(vals)
    ax.bar(x[mask] + offset, vals[mask], width=width, label=year)


ax.set_ylabel("Price (USD)")
ax.set_title("Average Retail Product Price Per Year")
ax.set_xticks(x)
ax.set_xticklabels(products)
ax.set_ylim(0, 500)
ax.legend(title="Year", bbox_to_anchor=(1.05, 1), loc="upper left")
ax.grid(axis="y", linestyle="--", linewidth=0.5)


import matplotlib.ticker as mtick
ax.yaxis.set_major_formatter(mtick.StrMethodFormatter("${x:,.0f}"))

plt.tight_layout()
plt.show()

```
